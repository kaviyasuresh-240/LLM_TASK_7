## Sentiment Visualization
The following chart shows the distribution of sentiment labels detected across the article sentences. 
It provides a quick overview of the emotional tone of the text.


```python
import os
os.chdir("D:/HUGGING FACE PIPELINES")
import pandas as pd
import matplotlib.pyplot as plt

# Load sentiment results
sentiment_df = pd.read_csv("dataset/sentiment_results.csv")

# Count sentiment labels
sentiment_counts = sentiment_df['label'].value_counts()

# Plot sentiment distribution
plt.figure(figsize=(6,4))
sentiment_counts.plot(kind='bar', color=['skyblue', 'lightgreen', 'salmon'])
plt.title("Sentiment Distribution Across Article Sentences")
plt.xlabel("Sentiment Label")
plt.ylabel("Count")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

### Interpretation of Sentiment Visualization
The bar chart above illustrates the emotional tone detected across the article sentences.  
Most sentences show a **positive** or **neutral** sentiment, reflecting the informative and optimistic nature of the healthcare topic.  
This distribution indicates that the article emphasizes progress, innovation, and ethical awareness rather than criticism or negativity.  
Such sentiment patterns are typical of analytical or educational texts that aim to inform readers rather than evoke strong emotions.
